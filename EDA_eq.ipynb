```python
# import necessary libraries
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import polars as pl
```

```python
## load in datasets - take in genus-level data

# dataset for disease vs. healthy household pair study
zhou_meta = pl.read_excel('data/Table_S1_Participant_and_samples.xlsx', 
                        sheet_name='Sample phenotype').to_pandas()
zhou_abundances = pl.read_excel('data/Table_S5_Metagenomics_data.xlsx', 
                        sheet_name='genus').to_pandas()

# Israeli cohort dataset
levi_meta = pl.read_excel('data/mmc3.xlsx', sheet_name='Microbiome particpants').to_pandas()
levi_abundances = pl.read_excel('data/mmc4.xlsx', 
                        sheet_name='Species').to_pandas() # the column names have the genus info in it
```

## EDA of Zhou datasets

Paper: [*Gut microbiome of multiple sclerosis patients and paired household healthy controls reveal associations with disease risk and course* (Zhou et al, 2023)](https://pmc.ncbi.nlm.nih.gov/articles/PMC10143502/)

### Metagenomics Abundance Data

```python
# species-level abundance data
num_taxa = zhou_abundances.shape[0]
num_samples = zhou_abundances.shape[1] - 1 # one column is Taxa

print(f"Number of taxa: {num_taxa}")
print(f"Each row (species-level taxa) is unique: {zhou_abundances['Taxa'].nunique() == num_taxa}")
print(f"Number of samples in abundance data: {num_samples}")
zhou_abundances.head()
```

### Patient-Level Metadata

```python
# metadata
print(f"Number of samples in metadata: {zhou_meta.shape[0]}")
print(f"Columns in metadata: {zhou_meta.columns}")
zhou_meta.head()
```

```python
# look at only the metadata features for disease/treatment status
disease_cols = zhou_meta.loc[:,zhou_meta.columns.str.contains('disease$|disease_course|iMSMS_ID|treatment_status')].copy()
disease_cols.head()
```

### Further EDA of the Abundance + Metadata

See if the sample IDs match between the abundance matrix and the metadata:

```python
# see if the sample IDs match between the abundance matrix and the metadata
abundance_sampleids = set(zhou_abundances.columns[1:])
meta_sampleids = set(zhou_meta['iMSMS_ID'])

sample_matches = abundance_sampleids.intersection(meta_sampleids)
print(f"Number of matching samples between abundance and metadata: {len(sample_matches)}")
print(f"Match between sample IDs in abundance and metadata: {zhou_abundances.shape[1] - 1 == len(sample_matches)}")
```

Check for missing values

```python
def check_missing(abundance_df: pd.DataFrame, meta_df: pd.DataFrame, disease_df: pd.DataFrame):
    missing_abundance_samples = abundance_df[abundance_df.isna().any(axis=1)]
    print(f"Any missing values by-sample in abundance data: {len(missing_abundance_samples)}")

    missing_meta_samples = meta_df[meta_df.isna().any(axis=1)]
    print(f"Any missing values by-sample in metadata: {len(missing_meta_samples)}")

    missing_disease_samples = disease_df[disease_df.isna().any(axis=1)]
    print(f"Any samples missing disease or treatment status information: {len(missing_disease_samples)}")

check_missing(zhou_abundances, zhou_meta, disease_cols)
```

```python
# overview of missing values in metadata
zhou_meta.isna().sum()
```

Check for unique values

```python
disease_status = disease_cols['disease'].unique()
treatment_status = disease_cols['treatment_status'].unique()
print(f"Unique disease status values: {disease_status}")
print(f"Unique treatment status values: {treatment_status}")

num_treated = len(disease_cols[disease_cols['treatment_status'] == 'Treated'])
num_untreated = len(disease_cols[disease_cols['treatment_status'] == 'Untreated'])
num_control = len(disease_cols[disease_cols['treatment_status'] == 'Control'])
print(f"Number of treated MS patients: {num_treated}")
print(f"Number of untreated MS patients: {num_untreated}")
print(f"Number of control patients: {num_control}")
```

Look for features (taxa) that have zero abundances across all samples

```python
# sum across all the samples (columns) for each feature
zero_abundance_features = zhou_abundances.index[zhou_abundances.sum(axis=1, numeric_only=True) == 0]
print(f"Number of taxa which have zero abundances across all samples: {len(zero_abundance_features)}")
```

Look at how many of each disease type:

```python
disease_types = disease_cols['disease_course'].unique()
print(f"Unique MS types: {disease_types}")
disease_cols['disease_course'].value_counts()
```

#### Conclusions:

- There are 546 total genus-level taxa in this dataset, with no duplicates.
- There are no taxa which have completely zero abundance.
- There are 1152 sample in total.
- We have an exact match of all 1152 samples in the abundance data to the sample IDs in the metadata.
- There are no missing abundance values for any of the samples.
- There are no missing metadata related to disease status or treatment status.
- Each patient is either MS/Control (n=576 each). Within MS patients, they are either treated (n=367) or untreated (n=209).
- Among MS patients these were their MS disease types: 437 are RRMS, 71 are PPMS, and 68 are SPMS.

## EDA of Levi datasets

Paper: [*Potential role of indolelactate and butyrate in multiple sclerosis revealed by integrated microbiome-metabolome analysis* (Levi et al, 2021)](https://pmc.ncbi.nlm.nih.gov/articles/PMC8080254/)

### Metagenomics Abundance Data

```python
# species-level abundance data
num_taxa = levi_abundances.shape[1] - 1 # one column is for sample ID
num_samples = levi_abundances.shape[0] 

# rename the SampleID column
levi_abundances.rename(columns={'__UNNAMED__0': 'SampleID'}, inplace=True)

print(f"Number of taxa: {num_taxa}")
print(f"Number of samples in abundance data: {num_samples}")
levi_abundances.head()
```

### Patient-Level Metadata

```python
# metadata
print(f"Number of samples in metadata: {levi_meta.shape[0]}")
print(f"Columns in metadata: {levi_meta.columns}")

# join together Registrant_number and FD columns to get a unique ID that matches the abundance data
levi_meta['SampleID'] = levi_meta['Registrant_number'].astype(str) + '_' + levi_meta['FD'].astype(str)
levi_meta.head()
```

```python
# look at only the metadata features for disease/treatment status
disease_cols = levi_meta.loc[:,levi_meta.columns.str.contains('MS_type short|treatment_group|SampleID')].copy()
disease_cols.head()
```

### Further EDA of the Abundance + Metadata

See if the sample IDs match between the abundance matrix and the metadata:

```python
# see if the sample IDs match between the abundance matrix and the metadata
abundance_sampleids = set(levi_abundances['SampleID'])
meta_sampleids = set(levi_meta['SampleID'])

sample_matches = abundance_sampleids.intersection(meta_sampleids)
print(f"Number of matching samples between abundance and metadata: {len(sample_matches)}")
print(f"Match between sample IDs in abundance and metadata: {levi_abundances.shape[0] == len(sample_matches)}")
```

Check for missing values

```python
def check_missing(abundance_df: pd.DataFrame, meta_df: pd.DataFrame, disease_df: pd.DataFrame):
    missing_abundance_samples = abundance_df[abundance_df.isna().any(axis=1)]
    print(f"Any missing values by-sample in abundance data: {len(missing_abundance_samples)}")

    missing_meta_samples = meta_df[meta_df.isna().any(axis=1)]
    print(f"Any missing values by-sample in metadata: {len(missing_meta_samples)}")

    missing_disease_samples = disease_df[disease_df.isna().any(axis=1)]
    print(f"Any samples missing disease or treatment status information: {len(missing_disease_samples)}")

check_missing(levi_abundances, levi_meta, disease_cols)
```

```python
# overview of missing values in metadata
levi_meta.isna().sum()
```

Check for unique values

```python
treatment_status = disease_cols['treatment_group'].unique()
disease_cols['treatment_group'] = disease_cols['treatment_group'].fillna('Control')
print(f"Unique treatment status values: {treatment_status}")

num_treated = len(disease_cols[disease_cols['treatment_group'] == 'MS treated'])
num_untreated = len(disease_cols[disease_cols['treatment_group'] == 'MS untreated'])
num_control = len(disease_cols[disease_cols['treatment_group'] == 'Control'])
print(f"Number of treated MS patients: {num_treated}")
print(f"Number of untreated MS patients: {num_untreated}")
print(f"Number of control patients: {num_control}")
```

Look for features (taxa) that have zero abundances across all samples

```python
# sum across all the samples (rows) for each feature
zero_abundance_features = levi_abundances.columns[1:][levi_abundances.sum(axis=0, numeric_only=True) == 0]
print(f"Number of taxa which have zero abundances across all samples: {len(zero_abundance_features)}")
```

Look at how many of each disease type:

```python
disease_types = disease_cols['MS_type short'].unique()
print(f"Unique MS types: {disease_types}")
disease_cols['MS_type short'].value_counts()
```

#### Conclusions:

- There are 686 total species-level taxa in this dataset, with no duplicates.
- There are no taxa which have completely zero abundance.
- There are 187 samples in total.
- We need to combine the Registrant number with FD column value to get the Sample ID in the metadata to get an exact match of all 187 samples in the abundance data to the sample IDs in the metadata.
- There are no missing abundance values for any of the samples.
- There are no missing metadata related to disease status or treatment status (the 58 NA values in the `treatment_group` columns are actually just Control samples, not real missing values).
- Each patient is either MS (n=129) or Control (n=58). Within MS patients, they are either treated (n=64) or untreated (n=65).
- Among MS patients these were their MS disease types: 51 are RRMS, 39 CIS, 25 are relapse, 8 are progressive, and 6 are RIS.

### Check if Feature Set is Transferable Between Datasets

```python
# get features in Zhou dataset
zhou_taxa = zhou_abundances['Taxa']
zhou_taxa
```

```python
# see features in Levi dataset
levi_taxa = pd.Series(levi_abundances.columns[1:])
levi_taxa[0]
```

```python
def get_species(feature_name):
    """Extract the g__ segment from Levi dataset to only get genus."""
    g__split = feature_name.split("g__")[1]
    return g__split.split('|')[0]

def clean_feature_name(feature_name):
    """Standardize feature names to be lowercase, and remove extra whitespace"""
    feature_lower = feature_name.lower().strip()
    return feature_lower

# get only genus name of Levi dataset
levi_species = levi_taxa.apply(get_species)

# standardize the features for both datasets
zhou_clean_features = zhou_taxa.apply(clean_feature_name)
levi_clean_features = levi_species.apply(clean_feature_name)

# get only unique features
zhou_unique_features = set(zhou_clean_features)
levi_unique_features = set(levi_clean_features)

# check exact genus-level overlap
exact_overlap = zhou_unique_features.intersection(levi_clean_features)
print(f"Number of Zhou dataset unique genus: {len(zhou_unique_features)}")
print(f"Number of Levi dataset unique genus: {len(levi_unique_features)}")
print(f"Exact matches: {len(exact_overlap)}")
print(f"{len(exact_overlap)/len(zhou_unique_features)}")
print(exact_overlap)
```
